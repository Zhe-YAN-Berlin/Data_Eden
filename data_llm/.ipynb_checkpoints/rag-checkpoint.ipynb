```python
import minsearch
```

```python
import json
```

```python
with open('llm-radar-content.json', 'rt') as f_in:
    docs_raw = json.load(f_in)
```

```python
#docs_raw
```

```python
documents = []
for course_dict in docs_raw:
    for doc in course_dict['documents']:
        doc['course'] = course_dict['course']
        documents.append(doc)
```

```python
documents[0]
```

```python
index = minsearch.Index(
    text_fields=["question", "text", "section"],
    keyword_fields=["course"]
)
```

```python
q = 'why power supply always important for the radar systems?'
```

```python
index.fit(documents)
```

```python
boost = {'question': 3.0, 'section':0.5}

results = index.search(
    query=q,
    filter_dict = {'course':'radar-maintenance'},
    boost_dict=boost,
    num_results = 5
)
```

```python
results
```

```python
def search(query):
    boost = {'question': 3.0, 'section':0.5}

    results = index.search(
        query = query,
        filter_dict = {'course':'radar-maintenance'},
        boost_dict=boost,
        num_results=5
    )
    return results
```

```python
def build_prompt(query, search_results):
    prompt_template = """"
You're a course teaching assitant in military to help young technical soliders to understand radar system better. Answer the QUESTION based on the CONTEXT.
Use only the facts from the CONTEXT when asnwering the QUESTION.
If the CONTEXT doesn't contain the answer, output NONE

QUESTION: {question}

CONTEXT: {context}
""".strip()
    context = ""

    for doc in search_results:
        context += (
            f"section: {doc['section']}\n"
            f"question: {doc['question']}\n"
            f"answer: {doc['text']}\n\n"
        )
    prompt = prompt_template.format(question=q, context=context).strip()
    return prompt
```

```python
def llm(prompt):
    response = client.chat.completions.create(
    messages=[{"role":"user", "content":prompt}],
    model="gpt-4o-mini",
    )
    return response.choices[0].message.content
```

```python
##########
###test###
query = 'how to improve power supply?'
search_results = search(query)
prompt = build_prompt(query, search_results)
answer = llm(prompt)
```

```python
print(answer)
```

```python
##########
###test2##
query = 'which component of radar system is most crucial?'
def rag(query):
    search_results = search(query)
    prompt = build_prompt(query, search_results)
    answer = llm(prompt)
    return answer 
```
